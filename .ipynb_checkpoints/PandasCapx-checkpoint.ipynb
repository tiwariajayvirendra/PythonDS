```python
#String
import numpy as np
import pandas as pd
country=["india","Pakistan","Srilanka","Nepal"]

pd.Series(country)
```

```python
#integers
runs= [13,24,56,78,100]
pd.Series(runs)
```

```python
#custom index
marks=[67,45,67,78,100]
subjects=["Maths","ENGLISH","HINDI","GUJRATI","SCINECE"]

pd.Series(marks,index=subjects)
```

```python
#Setting a Name
marks=pd.Series(marks,index=subjects,name="Ajay Ke marks")
marks
```
