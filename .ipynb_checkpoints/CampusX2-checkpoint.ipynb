```python
lower = int(input("Enter Lower range"))
upper = int(input("Enter upper range"))

for i in range(lower,upper+1):
    for j in range (2,i):
        if i % j == 0:\
            
            break
    else:
        print(i)
    
```

```python
#Continue  -> current itteration ko skip kar deta hai ;
for i in range (1,10):
    if i==5:
        continue
    print(i)
# for an real world example we use continue in case of product based company were product stocks are 0 so you don't want to show it on the page

```

```python
# while using Negative Indexing first number will be bigger in comparision of 2nd number 
# In this case we will see -1 is bigger then -6  because it is negative counting 

s = 'hello world'
print(s[-1:-6:-1])

```
