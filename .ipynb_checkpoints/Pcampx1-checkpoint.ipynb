```python
#Problem 1: Write a program that will give you in hand monthly salary after 
#deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:
# Salary(Lakhs) : Tax(%)

# Below 5 : 0%
# 5-10 : 10%
# 10-20 : 20%
# aboove 20 : 30%
#SOlUTION 

#Actual Annual salay in Lack

salary=int(input("Enter Total Salary:"))

#Deduction Percentage

hra= (salary * 10) / 100
da= (salary * 5) / 100
pf= (salary * 3) / 100

print("HRA:",hra)
print("DA:",da)
print("PF:",pf)



#Tax Slab /
if salary < 500000:
    tax=0
elif salary <= 1000000:
    tax= (salary * 10)/100
elif salary <= 2000000:
    tax = (salary * 20)/100
else:
    tax = (salary * 30)/100
# Total Deductionable Tax
print("Tax:",tax)

#Total Deduction

total_deduction = tax+hra+da+pf
net_annual = salary -total_deduction
monthly_inhand = net_annual/12

#Print

print("Total Deduction:",total_deduction)
print("Net Annual salary:",net_annual)
print("Monthly In-Hand Salary:",monthly_inhand)


    
```
