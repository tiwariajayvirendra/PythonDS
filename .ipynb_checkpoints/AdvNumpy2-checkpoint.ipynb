```python
#Brodcasting Very Important Topice 
import numpy as np
a =np.arange(6).reshape(2,3)
b =np.arange(6,12).reshape(2,3)
print(a,"a")
print(b,"b")

print(a+b)
```

```python
#Sigmoid
#mean Squared error
#categorical cross entrog
def sigmoid(array):
    return  1/(1 + np.exp(-(array)))
a=np.arange(100)
sigmoid(a)
```
