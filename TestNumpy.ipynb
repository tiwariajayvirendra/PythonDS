```python
# Q-2 Ask user to input two numbers a, b. Write a program to generate a random array of shape (a, b) and print the array and avg of the array.
a = int(input("Enter Number Of rows(a): "))
b = int(input("Enter Number of Colums(b): "))

array = np.random.rand(a,b) #Shape (a,b) Random array
avg = np.mean(array)

print("Random array :")
print(array)
print("Average of array:",avg)
```

```python
#Numpy array vs python Lists\
#Speed
#list
import time
a = [i for i in range(10000000)]
b = [ i for i in range(10000000,200000000)]

c=[]

start = time.time()
for i in range(len(a)):
    c.append(a[i]+b[i])
print(time.time()-start)
```

```python
#Numpy 
import numpy as np
a = np.arange(10000000)
b = np.arange(10000000,20000000)
start = time.time()
c = a + b
print(time.time()-start)
```

```python
# Numpy Array is more faster then python List 
```

```python
#Built in python data type List 
# Terms of Memory 
a = [i for i in range(10000000)]
import sys

sys.getsizeof(a)
```

```python
#Numpy 
a = np.arange(10000000, dtype=np.int8)
sys.getsizeof(a)
```
