```python
class animal:
    def sound(self):
        print("make sound:")
class dog(animal):
    def sound(self):
        super.sound()
        print("dog bark")
class cat (animal):
    def sound(self):
        super.sound()
        print("cat meow")
animalss=dog()
animals2=cat()
```
